```python
import matplotlib.pyplot as plt
import numpy as np
import textwrap
```

```python
bit_8 = list(range(0,256))
bit_14 = list(range(0,16384))
bit_16 = list(range(0,65536))
```

```python
print(bit_8)
```

```python
max(bit_16)
```

```python
norm_8 = [0 if i==0 else i/255 for i in bit_8]
norm_14 = [0 if i==0 else i/16383 for i in bit_14]
norm_16 = [0 if i==0 else i/65535 for i in bit_16]
```

```python
print(norm_8)
```

```python
plt.figure(figsize=(30, 20))

plt.plot(norm_8, norm_8, drawstyle='steps-post', label='8-bit (256 gradations)', linewidth=4)
plt.plot(norm_14, norm_14, drawstyle='steps-post', label='14-bit (16384 gradations)', linewidth=4)
plt.plot(norm_16, norm_16, drawstyle='steps-post', label='16-bit (65536 gradations)', linewidth=4)

# Add diagonal reference line (perfect continuous mapping)

plt.xlabel('Normalized Gray Value',fontsize=24,weight='bold',labelpad=30)
plt.ylabel('Normalized Gray Value',fontsize=24,weight='bold',labelpad=30)
plt.title('Comparison of Bit Depths',fontsize=36,weight='bold',pad=40)
plt.legend(fontsize=22,markerscale=4)
plt.grid(True, alpha=0.3)  # Make sure both axes have the same scale
plt.xlim(0.998, 1)
plt.ylim(0.998, 1)
plt.xticks(fontsize=18)
plt.yticks(fontsize=18)
plt.savefig(r'C:\Users\Kristin.Gallik\OneDrive - Van Andel Institute\Workshops_Seminars\Experimental_Skills_2025\bit_depth_comparison_0998-1.png', dpi=300)
plt.show()
```
