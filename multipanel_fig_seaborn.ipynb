```python
import skimage as sk
import os
from glob import glob
import re
import random
import matplotlib.pyplot as plt
```

```python
# get all images that have paired detection overlays
vai_files = sorted(glob('/home/gkristin/Desktop/Trick_vial_manus/Figures/Raw_Images_for_Figs/VAI_Images/*.png'))

a6 = []
b5 = []
c1 = []
c4 = []
d3 = []

for file in vai_files:
    if 'A6' in os.path.basename(file):
        a6.append(file)
    elif 'B5' in os.path.basename(file):
        b5.append(file)
    elif 'C1' in os.path.basename(file):
        c1.append(file)
    elif 'C4' in os.path.basename(file):
        c4.append(file)
    elif 'D3' in os.path.basename(file):
        d3.append(file)
    else:
        print(f'{os.path.basename(file)} not matching pattern search, skipping')
```

```python
def get_imgs_with_detections(all_files):
    files = []
    for file in all_files:
        if 'A6' in os.path.basename(file):
            files.append(file)
        elif 'B5' in os.path.basename(file):
            files.append(file)
        elif 'C1' in os.path.basename(file):
            files.append(file)
        elif 'C4' in os.path.basename(file):
            files.append(file)
        elif 'D3' in os.path.basename(file):
            files.append(file)
        else:
            print(f'{os.path.basename(file)} not matching pattern search, skipping')
    return files   

def sort_imgs_detections(file_list):
    detections = []
    imgs = []
    for file in file_list:
        if 'Detections' in os.path.basename(file):
            detections.append(file)
            #print(f'Added {file} to detection list')
        else:
            imgs.append(file)
            #print(f'Added {file} to image list')
    return detections, imgs

def random_subset(detections, imgs, n):
    random_selection_detections = random.sample(detections,n)
    print(f'finding matches for: {list(map(os.path.basename,random_selection_detections))}')
    random_selection_imgs = []
    for sample in random_selection_detections:
        name = re.search(r'[0-9]{6}_[A-z][0-9]', os.path.basename(sample)).group()
        print(f'looking for {name}')
        img_match = next(img for img in imgs if name in os.path.basename(img))
        random_selection_imgs.append(img_match)
        print(f'found {name} and added to list')
    return random_selection_imgs, random_selection_detections


def random_selection(detections, n):
    a6 = [file for file in detections if 'A6' in file]
    b5 = [file for file in detections if 'B5' in file]
    c1 = [file for file in detections if 'C1' in file]
    c4 = [file for file in detections if 'C4' in file]
    d3 = [file for file in detections if 'D3' in file]
    names = []
    names.extend(random.sample(a6,n))
    names.extend(random.sample(b5,n))
    names.extend(random.sample(c1,n))
    names.extend(random.sample(c4,n))
    names.extend(random.sample(d3,n))
    return names

def save_selection_to_file(names,path):
    path = os.path.join(path,'randomly_selected_images.txt')
    with open(path, 'w') as file:
        for name in names:
            base_name = os.path.basename(name)
            file.write(f'{os.path.basename(base_name)}\n')
```

```python
path = '/home/gkristin/Desktop/Trick_vial_manus/Figures/Raw_Images_for_Figs/Collab_Images/'
collab_files = sorted(glob('/home/gkristin/Desktop/Trick_vial_manus/Figures/Raw_Images_for_Figs/Collab_Images/*.png'))
files = get_imgs_with_detections(collab_files)
detections, _ = sort_imgs_detections(files)
names = random_selection(detections,3)
save_selection_to_file(names,path)
```

```python
detections, imgs = sort_imgs_detections(a6)
random_selection_imgs, random_selection_detections = random_subset(detections,imgs,3)
```

```python
a6_imgs = list(map(sk.io.imread,random_selection_imgs))
a6_detections = list(map(sk.io.imread,random_selection_detections))
```

```python
all_imgs = a6_imgs + a6_detections
```

```python
fig, axs = plt.subplots(2,3,subplot_kw={'xticks':[], 'yticks': []})
for img, ax in zip(all_imgs,axs.flat):
    ax.imshow(img)
plt.tight_layout()
plt.show()
```
