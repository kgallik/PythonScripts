```python
import numpy as np
import skimage as sk
import os
import napari
```

```python
img = "E:/Steensma_Lab/Abigail/Raw_ROI.jpg"
tiles = "E:/Steensma_Lab/Abigail/Pixel_Classifier_Overlay_ROI.jpg"
```

```python
img = sk.io.imread(img)
tiles = sk.io.imread(tiles)
```

```python
def get_gradient_2d(start, stop, width, height, is_horizontal):
    if is_horizontal:
        return np.tile(np.linspace(start, stop, width), (height, 1))
    else:
        return np.tile(np.linspace(start, stop, height), (width, 1))
```

```python
img.dtype
```

```python
gradient = get_gradient_2d(0,256,width=img.shape[1],height=img.shape[0],is_horizontal=True)
```

```python
viewer = napari.view_image(gradient)
```

```python
mask = sk.transform.resize(gradient, img.shape)
```

```python
mask = mask/255
```

```python
blend = img * mask + tiles*(1-mask)
```

```python
viewer = napari.view_image(blend)
```

```python
blend_rgb = sk.color.convert_colorspace(blend, 'RGB', 'RGB')
```

```python
blend = np.astype(blend, np.uint8)
```

```python
sk.io.imsave('E:/Steensma_Lab/Abigail/Blended.jpg',blend)
```
